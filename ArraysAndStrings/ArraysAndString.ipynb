# Two Pointers

```python
#392. Is Subsequence
#Given two strings s and t, return true if s is a subsequence of t, or false otherwise.
def isSubsequence(pattern, text):
    """
    :type pattern: str
    :type text: str
    :rtype: bool
    """
    matched = 0
    for char in text:
        if matched < len(pattern) and pattern[matched] == char:
            matched += 1
    return matched == len(pattern)
#Time: O(n) -> Space: O(1)
```

```python
s = "abc"
t = "ahbgdc"
print("Should be Output: true -> ", isSubsequence(s, t))
```

```python
s = "axc"
t = "ahbgdc"
print("Should be Output: false -> ", isSubsequence(s, t))
```
