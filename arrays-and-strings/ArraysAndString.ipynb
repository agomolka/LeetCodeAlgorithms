# Two Pointers

```python
#392. Is Subsequence
#Given two strings s and t, return true if s is a subsequence of t, or false otherwise.
def isSubsequence(pattern, text):
    """
    :type pattern: str
    :type text: str
    :rtype: bool
    """
    matched = 0
    for char in text:
        if matched < len(pattern) and pattern[matched] == char:
            matched += 1
    return matched == len(pattern)
#Time: O(n) -> Space: O(1)
```

```python
s = "abc"
t = "ahbgdc"
print("Should be Output: true -> ", isSubsequence(s, t))
```

```python
s = "axc"
t = "ahbgdc"
print("Should be Output: false -> ", isSubsequence(s, t))
```

# Reverse String
Write a function that reverses a string. The input string is given as an array of characters s.

You must do this by modifying the input array in-place with O(1) extra memory.

```python
class Solution(object):
    def reverseString(self, s):
        """
        :type s: List[str]
        :rtype: None
        Do not return anything, modify s in-place instead.
        """

        left = 0
        right = len(s) - 1

        while left < right:
            s[left], s[right] = s[right], s[left]
            left += 1
            right -= 1


def check_test(input_array, expected):
    original = input_array[:]

    solution = Solution()
    solution.reverseString(input_array)

    print(
        f"input={original} | expected={expected} | "
        f"got={input_array} | passed={input_array == expected}"
    )
```

```python
check_test(["h", "e", "l", "l", "o"], ["o", "l", "l", "e", "h"])
check_test(["H", "a", "n", "n", "a", "h"], ["h", "a", "n", "n", "a", "H"])
```
