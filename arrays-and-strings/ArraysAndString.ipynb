# Two Pointers

```python
#392. Is Subsequence
#Given two strings s and t, return true if s is a subsequence of t, or false otherwise.
def isSubsequence(pattern, text):
    """
    :type pattern: str
    :type text: str
    :rtype: bool
    """
    matched = 0
    for char in text:
        if matched < len(pattern) and pattern[matched] == char:
            matched += 1
    return matched == len(pattern)
#Time: O(n) -> Space: O(1)
```

```python
s = "abc"
t = "ahbgdc"
print("Should be Output: true -> ", isSubsequence(s, t))
```

```python
s = "axc"
t = "ahbgdc"
print("Should be Output: false -> ", isSubsequence(s, t))
```

# Reverse String
Write a function that reverses a string. The input string is given as an array of characters s.

You must do this by modifying the input array in-place with O(1) extra memory.

```python
#Time:O(n) Space: O(1)
class Solution(object):
    def reverseString(self, s):
        """
        :type s: List[str]
        :rtype: None
        """

        left = 0
        right = len(s) - 1

        while left < right:
            s[left], s[right] = s[right], s[left]
            left += 1
            right -= 1


def check_test(input_array, expected):
    original = input_array[:]

    solution = Solution()
    solution.reverseString(input_array)

    print(
        f"input={original} | expected={expected} | "
        f"got={input_array} | passed={input_array == expected}"
    )
```

```python
check_test(["h", "e", "l", "l", "o"], ["o", "l", "l", "e", "h"])
check_test(["H", "a", "n", "n", "a", "h"], ["h", "a", "n", "n", "a", "H"])
```

# Squares of a Sorted Array

Given an integer array nums sorted in non-decreasing order, return an array of the squares of each number sorted in non-decreasing order.

```python
#O(n log n)
def sortedSquares(nums):
    """
    :type nums: List[int]
    :rtype: List[int]
    """
    n = len(nums) #get the len of nums
    result = [0] * n #create an output array filled with zeros
    
    left = 0 #start one pointer at the beginning of nums
    right = n - 1 #start second pointer at the end of nums
    position = n - 1 #start filling result from the last index
    
    while left <= right: #continue while there are still numbers between left and right
        left_square = nums[left] * nums[left] #square the number at the left pointer
        right_square = nums[right] * nums[right] #square the number at the right pointer
        
        if left_square > right_square: #if the left square is bigger -> belongs at the current largest open position
            result[position] = left_square #put the bigger square into the current largest
            left += 1 #move the left pointer to the right, because nums[left] has now been used
        else: #the right square is bigger or equal
            result[position] = right_square #put the right pointer into the current position
            right -= 1 #move the right pointer to the left because nums[right] has now been used
        position -= 1 #move the next position from the end
    return result #return the fully filled sorted array
```

```python
def check_test(nums, expected):
    result = sortedSquares(nums)
    print(f"nums={nums} | expected={expected} | "f"got={result} | passed={result == expected}")

check_test([-4, -1, 0, 3, 10], [0, 1, 9, 16, 100])
check_test([-7, -3, 2, 3, 11], [4, 9, 9, 49, 121])
```

# Sliding window
Given an array of positive integers nums and an integer k, find the length of the longest subarray whose sum is less than or equal to k. This is the problem we have been talking about above. We will now formally solve it.

```python
#Time:O(n) Space: O(1)
def longest_subarray_sum_at_most_k(nums, k):
    left = 0 #left is the start index
    current = 0 #stores the sum of the current window nums[left:right+1]
    best = 0 #stores the longest valid window length found so far
    for right in range(len(nums)): #move right from the beginning to the end of the array
        current += nums[right] #add the new right-side element into the window
        while current > k: #if the current window sum is too large, shrink from the left
            current -= nums[left] #remove nums[left] from the current window sum
            left += 1 #move left forward because nums[left] is no longer inside the window
        best = max(best, right - left + 1) #curr <= k, so the window is valid
    return best #the best length found
```

```python
nums = [3, 1, 2, 7, 4, 2, 1, 1, 5]
k = 8
print(longest_subarray_sum_at_most_k(nums, k))
```

# Sliding window
You are given a binary string s (a string containing only "0" and "1"). You may choose up to one "0" and flip it to a "1". What is the length of the longest substring achievable that contains only "1"?

For example, given s = "1101100111", the answer is 5. If you perform the flip at index 2, the string becomes 1111100111.

```python
#Time:O(n) Space: O(1)
def longest_ones_after_one_flip(s):
    left = 0 #start index
    zero_count = 0 #tracks how many zeros are inside the current window
    best = 0 #stored the longest valid window
    for right in range(len(s)): #moves through the string
        if s[right] == "0": #if the char is "0" the window now has one more zeros
            zero_count += 1
        while zero_count > 1: #shrink from left, is we have more than one zero
            if s[left] == "0": #if the char leaving the window is "0" -> window loses zero
                zero_count -= 1
            left += 1 #move the left coundary
        best = max(best, right - left + 1) #at this point the window has at most one zero -> can become all ones after the most one flip
    return best #longest valid window length
```

```python
s = "1101100111"
longest_ones_after_one_flip(s)
```

```python
s = "1111"
longest_ones_after_one_flip(s)
```

# Sliding window
Example 3: 713. Subarray Product Less Than K.

Given an array of positive integers nums and an integer k, return the number of subarrays where the product of all the elements in the subarray is strictly less than k.

```python
def num_subarray_product_less_than_k(nums, k):
    if k <= 1: #no positive product can be strictly less than k
        return 0

    left = 0 #start index
    product = 1 #start with 1, because 1 is the neutral value for multiplication
    count = 0 #stores the number of valid subarrays found
    
    for right in range(len(nums)): #expand the number of valid subarrays found
        product *= nums[right] #include nums[right] in the current window
        
        while product >= k: #if the product is too large shrink the window from the left
            product //= nums[left] #remove nums[left] from the product because it leaves the window
            left += 1 #move left boundary
        count += right - left + 1 #now product<k, so the current window is valid

    return count
```

```python
num_subarray_product_less_than_k([10, 5, 2, 6], 100)
```
