```python
#392. Is Subsequence
#Given two strings s and t, return true if s is a subsequence of t, or false otherwise.

from collections import defaultdict
import bisect

class SubsequenceChecker(object):
    def __init__(self, t):
        self.positions = defaultdict(list)
        for i, char in enumerate(t):
            self.positions[char].append(i)
            
    def isSubsequence(self, s):
        prev_index = -1
        
        for char in s:
            if char not in self.positions:
                return False
            
            indexes = self.positions[char]
            
            pos = bisect.bisect_right(indexes, prev_index)
            
            if pos == len(indexes):
                return False
            
            prev_index = indexes[pos]
            
        return True
```

```python
checker = SubsequenceChecker("ahbgdc")

result1 = checker.isSubsequence("abc")
expected1 = True
print(f'Test 1: s="abc", t="ahbgdc" | expected={expected1}, got={result1} | passed={result1 == expected1}')

result2 = checker.isSubsequence("axc")
expected2 = False
print(f'Test 2: s="axc", t="ahbgdc" | expected={expected2}, got={result2} | passed={result2 == expected2}')
```
